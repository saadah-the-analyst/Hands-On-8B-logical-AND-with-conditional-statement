# Lesson 8B - Logical AND with Conditional Statements (if, if...else and if...elif...else)
## Mini Project: Secure Employee Access Verification System

```python
/# -------------------------------------------------------------------
#           SecureTech Solutions - Access Control System 
# -------------------------------------------------------------------

# Step 1: Store the employee information (inputs)
employee_name = input("Enter employee name: ")
age = int(input("Enter age: "))
employment_status = input("Employment Status: ")
department = input("Department: ")
security_clearance_level = int(input("Security Clearance: "))

print() # Blank line for clean spacing

# Step 2: Verify all security requirements using logical AND
# Core Business Rules:
# 1. Be 18 years or older (age >= 18)
# 2. Be an Active employee (employment_status == "Active")
# 3. Belong to IT or Cybersecurity department
# 4. Security clearance level of 3 or above (security_clearance_level >= 3)

is_age_valid = age >= 18
is_active = employment_status.strip().capitalize() == "Active"
is_valid_dept = department.strip() in ["IT", "Cybersecurity"]
is_clearance_valid = security_clearance_level >= 3

# Combine all conditions with logical AND
has_access = (
    is_age_valid and is_active and is_valid_dept and is_clearance_valid
)

# Steps 3 & 4: Determine access level and display appropriate message
if has_access:
    print(f"Employee: {employee_name}")
    print("Access Granted")

# Display security level based on clearance level using if...elif...else
if security_clearance_level == 3:
    print("Security Level: Standard Secure Access")
elif security_clearance_level == 4:
    print("Security Level: Advanced Secure Access")
elif security_clearance_level >= 5:
    print("Security Level: Top Secret Access")
else:
    print(f"Employee: {employee_name}")
    print("Access Denied")

```
